```python
# Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.
matrix=[
    [1,2,3],
    [4,5,6],
    [8,9,10]
]
def transpose_matrix(matrix):
    transpose=[]
    for i in range(3):
        row=[]
        for j in range(3):
            row.append(matrix[j][i])
        transpose.append(row)
    return transpose 
print("Original Matrix:")   
for row in matrix:
    print(row)

transposed=transpose_matrix(matrix)
print("\nTransposed Matrix")
for row in transposed:
    print(row)
```

```python
# Write a function that takes a nested list and flattens it into a single list. 
# Print the original and flattened lists.
nested=[
    [1,2,3],
    [4,5],
    [6,7,8]
]
def flatten_list(nested_list):
    flat_list=[]
    for row in nested_list:
        for item in row:
            flat_list.append(item)
    return flat_list

print("Original nested list:")
print(nested)
# flatten function call now
flattened=flatten_list(nested)
print("Flattened list:")
print(flattened)
```

```python
# Create a list of the first 10 positive integers. 
# Remove the elements at indices 2, 4, and 6, and insert the element '99' at index 5. 
# Print the modified list.

#different way to make list 
nums=list(range(1,11))
print("Original list:", nums)

# Remove elements at indices 2, 4, and 6 (remove in reverse order to avoid index shift)
# if we want to remove 7 and 6...if we remove 6 first then 7 index wala shift hoke change ho jayega
for index in sorted([2,4,6],reverse=True):
    nums.pop(index)

#Insert 9 at index 5
nums.insert(5,99)

print("Modified list", nums)
```

```python
# Write a function that takes a list and returns a new list with the elements in reverse order. 
# Print the original and reversed lists.
numbers=[10,20,30,40,50]
def reverse_list(lst):
    reversed_list=[]
    for i in range(len(lst)-1,-1,-1):
     reversed_list.append(lst)
    return reversed_list

print("Original list:", numbers)
# function call
reversed_numbers=reverse_list(numbers)
print(reversed_numbers)
```

```python
# Write a function that rotates a list by n positions. 
# Print the original and rotated lists.
def rotate_list(lst,n):
    n=n%len(lst)
    return lst[-n:]+lst[:-n]
numbers=[1,2,3,4,5]
n=2
rotated=rotate_list(numbers,n)
print(numbers)
print(rotated)
```
