```python
# Question 3
from sklearn.model_selection import train_test_split
data={
    "Study_Hours":[2,4,6,8,10],
    "Marks":[50,60,70,85,95]
}
df=pd.DataFrame(data)
x=df[["Study_Hours"]]
y=df["Marks"]
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=42
)
# Display shapes
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
# case study of student dataset
# Task 1: Load the Dataset
from sklearn.model_selection import train_test_split

df=pd.read_csv("student-perf.csv")
print(df.head())

# Task 2: Data Inspection
#dataset information
df.info()
print(df.isnull().sum())

# Task 3: Handle Missing Values (Correct ML Way)
# Numerical columns
num_imputer = SimpleImputer(strategy="mean")
df[["Age", "Final_Marks", "Attendance_Percentage"]] = num_imputer.fit_transform(
    df[["Age", "Final_Marks", "Attendance_Percentage"]]
)

# Categorical column
cat_imputer = SimpleImputer(strategy="most_frequent")
df[["Gender"]] = cat_imputer.fit_transform(df[["Gender"]])
# Task 4: Encode Categorical Data
df["Gender"] = df["Gender"].map({"Male": 0, "Female": 1})
# Task 5: Feature Selection
x = df[["Age", "Gender", "Attendance_Percentage"]]
y = df["Final_Marks"]
# Task 6: Data Discretization (Age into 3 uniform bins)
df["Age_Group"] = pd.cut(df["Age"], bins=3, labels=[0, 1, 2])
# Task 7: Train–Test Split (70% – 30%)
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=42
)

# Display shapes
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```
