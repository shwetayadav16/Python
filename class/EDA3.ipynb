```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.set(style="whitegrid")

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("train.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['SalePrice'], bins=30, kde=True)
plt.title("Distribution of House Prices")
plt.show()
```

```python
df['LogSalePrice'] = np.log(df['SalePrice'])

plt.figure(figsize=(8,5))
sns.histplot(df['LogSalePrice'], bins=30, kde=True)
plt.title("Log-Transformed SalePrice")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='OverallQual', data=df)
plt.title("Overall Quality Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='BedroomAbvGr', data=df)
plt.title("Bedroom Count Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='GrLivArea', y='SalePrice', data=df)
plt.title("Living Area vs Sale Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='BedroomAbvGr', y='SalePrice', data=df)
plt.title("Bedrooms vs Sale Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='OverallQual', y='SalePrice', data=df)
plt.title("Overall Quality vs Sale Price")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(x='Neighborhood', y='SalePrice', data=df)
plt.xticks(rotation=90)
plt.title("Neighborhood vs Sale Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='GrLivArea',
    y='SalePrice',
    hue='OverallQual',
    palette='viridis',
    data=df
)
plt.title("Area, Quality and Price")
plt.show()
```

```python
plt.figure(figsize=(12,8))

numeric_df = df.select_dtypes(include=['int64', 'float64'])
corr = numeric_df.corr()

sns.heatmap(corr, cmap='coolwarm', center=0)
plt.title("Correlation Heatmap (Numerical Features Only)")
plt.show()
```

```python
missing = df.isnull().mean().sort_values(ascending=False)
missing.head(10)
```

```python
plt.figure(figsize=(10,6))
missing[missing > 0].plot(kind='bar')
plt.title("Missing Value Percentage")
plt.show()
```
