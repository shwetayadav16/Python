```python
# 3. Writing to a File (w mode)
file=open("file.txt","r+")
file.write("Hello Python\n")
file.write("File handling")
file.seek(0)
data=file.read()
print(data)
```

```python
f=open("file.txt","a+")
f.write("New line added")
file.seek(0)
data=file.read()
print(data)
```

```python
# 6. Using with Statement (BEST PRACTICE)
# ✔ No need to write close()
# ✔ Used in real projects
with open("file.txt", "a+") as f:
    f.write("New line added")
    f.seek(0)
    print(f.read())
    f.close()
```

```python
# Check File Exists
import os
if os.path.exists("file.txt"):
    print("File exists")
else:
    print("File not found")
```
