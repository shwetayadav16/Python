```python
# 1️⃣ Write user input into a file
data=input()
with open("user_write.txt","w") as f:
    f.write(data)
```

```python
# 2️⃣ Count number of lines in a file
with open("file.txt","r") as f:
    count=0
    for line in f:
        count+=1
print(count)
```

```python
# 3️⃣ Count words in a file
with open("file.txt","r") as f:
    count=0
    for line in f:
        data=line.split()
        count+=len(data)
print(count)
```

```python
# 4️⃣ Copy data from one file to another
with open("file.txt","r") as f:
    with open("newfile.txt","w") as n:
        n.write(f.read())
        n.close()#need to close for reading
with open("newfile.txt","r") as n:
    print(n.read())
```

```python
# 5️⃣ Read file and print only even lines
with open("newfile.txt","r") as f:
    count=0
    for line in f:
        if count%2==0:
            print(line)
        count+=1
```
