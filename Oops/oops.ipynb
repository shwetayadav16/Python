```python
class BankAccount:
    def __init__(self,name,balance=0):
        self.name=name
        self.balance=balance
    # deposite
    def deposite(self,amount):
        if amount<=0:
            print("Deposit amount must be greater than 0")
            return
        self.balance+=amount
        print(f"Deposite ₹{amount}.New Balance:₹{self.balance}")
    # withdrawn
    def withdraw(self,amount):
        if amount<=0:
            print("Withdraw amount must be greater than 0")
            return
        if amount>self.balance:
            print("Insufficinet balance!")
            return 
        self.balance-=amount
        print(f"Withdrawn ₹{amount}. New Balance:₹{self.balance}")
    # withdrawn
    def check_balance(self):
        print(f"Current Balance:₹{self.balance}")

    # making object
acc1=BankAccount("Shweta",100000)
acc1.deposite(500)
acc1.withdraw(300)
acc1.check_balance()
```

```python
# Inheritance
##Single Inheritance
class father:
    def __init__(self,name,gender,age):
        self.name=name
        self.gender=gender
        self.age=age
    def nature(self):
            print("Anger Issue")
# Child class
class daughter(father):
    def __init__(self,name,gender,age,hobby):
        super().__init__(name,gender,age)#calling parent constructor
        self.hobby=hobby
    def nature(self):
        print("She is calm and polite")
        super().nature()#a way to create virtual function like in c++

#Object of Child class
d1=daughter("Shweta","Female",20,"Reading")
d1.nature()
print(d1.name,d1.gender,d1.age,d1.hobby)
```

```python
# Multiple Inheritance
class Father:
    def skills(self):
        print("Father: Driving")

class Mother:
    def skills(self):
        print("Mother: Cooking")

class Child(Father, Mother):
    def skills(self):
        Father.skills(self)
        Mother.skills(self)
        print("Child: Programming")

c1 = Child()
c1.skills()
```

```python
#Multiple Inheritance with __init__() (Best Practice using super())
class Father:
    def __init__(self):
        print("Father constructor called")
    def skills(self):
        print("Father: Driving")

class Mother:
    def __init__(self):
        print("Mother constructor called")
    def skills(self):
        print("Mother: Cooking")

class Child(Father, Mother):
    def __init__(self):
        super().__init__()   # only first parent in MRO runs
        print("Child constructor called")

c1 = Child()
c1.skills()
```

```python
class Father:
    def __init__(self):
        print("Father constructor called")
        super().__init__()

    def skills(self):
        print("Father: Driving")


class Mother:
    def __init__(self):
        print("Mother constructor called")
        super().__init__()

    def skills(self):
        print("Mother: Cooking")


class Child(Father, Mother):
    def __init__(self):
        print("Child constructor called")
        super().__init__()
    def skills(self):
        Father.skills(self)   # ✅ call Father skills
        Mother.skills(self)   # ✅ call Mother skills
c1 = Child()
c1.skills()
```
