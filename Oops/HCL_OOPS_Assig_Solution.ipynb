```python
# Question 1: Student Class
class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.roll_no)

student1 = Student("Rahul", 101)
student1.display()

```

```python
# Question 2: Simple Calculator
class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        print("Addition:", self.a + self.b)

    def subtract(self):
        print("Subtraction:", self.a - self.b)

calc = Calculator(10, 5)
calc.add()
calc.subtract()

```

```python
# Question 3: Bank Account
class BankAccount:
    def __init__(self, account_no, balance):
        self.account_no = account_no
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)
        print("Current Balance:", self.balance)

    def withdraw(self, amount):
        self.balance -= amount
        print("Withdrawn:", amount)
        print("Current Balance:", self.balance)

account = BankAccount(12345, 1000)
account.deposit(500)
account.withdraw(300)

```

```python
# Question 4: Rectangle
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        print("Area of Rectangle:", self.length * self.breadth)

rect = Rectangle(10, 5)
rect.area()
```

```python
 #Question 5: Constructor Practice
class Car:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def display(self):
        print("Brand:", self.brand)
        print("Price:", self.price)

car1 = Car("Toyota", 1500000)
car1.display()

```

```python
# Question 6: Inheritance
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    pass

dog = Dog()
dog.sound()

```

```python
# Question 7: Polymorphism
class Cat:
    def sound(self):
        print("Cat meows")

class Cow:
    def sound(self):
        print("Cow moos")

cat = Cat()
cow = Cow()
cat.sound()
cow.sound()

```

```python
# Question 8: __str__() Method
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"Book Title: {self.title}, Author: {self.author}"

book1 = Book("Python Basics", "Guido")
print(book1)

```

```python
# Question 9: Object Counter
class Counter:
    count = 0

    def __init__(self):
        Counter.count += 1

obj1 = Counter()
obj2 = Counter()
obj3 = Counter()
print("Total objects created:", Counter.count)

```

```python
# Question 10: Mobile
class Mobile:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def show(self):
        print("Brand:", self.brand)
        print("Price:", self.price)

mobile1 = Mobile("Samsung", 20000)
mobile2 = Mobile("Apple", 80000)
mobile1.show()
print()
mobile2.show()
```
